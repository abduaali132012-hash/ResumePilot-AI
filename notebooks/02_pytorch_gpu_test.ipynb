# 02 — PyTorch GPU Test

Prepares and verifies PyTorch on the AMD MI300X (ROCm).

> **Note:** on an AMD ROCm build, `torch.cuda` still appears — PyTorch keeps a
> CUDA-compatible device API abstraction over its GPU backend. So `True` here
> means the MI300X is being used, not that there is an NVIDIA GPU.

```python
import torch
print("PyTorch version:", torch.__version__)
```

```python
print("GPU available:", torch.cuda.is_available())
```

```python
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("Memory:", torch.cuda.get_device_properties(0).total_memory / (1024**3), "GiB")
else:
    print("Running on CPU.")
```

## GPU computation test

Not just "does PyTorch see the GPU?" — this actually asks the GPU to perform
a large matrix multiplication and measures the time.

```python
import time
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using:", device)

x = torch.randn(5000, 5000, device=device)
y = torch.randn(5000, 5000, device=device)

start = time.time()
z = torch.matmul(x, y)
if device == "cuda":
    torch.cuda.synchronize()
elapsed = time.time() - start

print("Result shape:", z.shape)
print("Time:", elapsed, "seconds")
```
