First model inference experiment, then moved onto the GPU.

Install (on the AMD machine):

Make sure PyTorch is the ROCm-compatible build supplied by the selected image.

````python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="distilgpt2",
)
````

````python
result = generator(
    "Artificial intelligence is",
    max_new_tokens=50,
)
print(result)
````

## Move it onto the GPU

Once everything works on CPU, point the pipeline at the MI300X:

````python
import torch

device = 0 if torch.cuda.is_available() else -1
print("device index:", device)

generator = pipeline(
    "text-generation",
    model="distilgpt2",
    device=device,
)
````

````python
result = generator(
    "Artificial intelligence is",
    max_new_tokens=50,
)
print(result)
````
